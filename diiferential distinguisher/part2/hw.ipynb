```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score

# Load your combined dataset
df = pd.read_csv("datasets/round_1_full.csv")

# -----------------------------------
# 1. Calculate Hamming weight
# -----------------------------------

bit_columns = [f"bit_{i}" for i in range(320)]

X_bits = df[bit_columns].astype(int)

df["hamming_weight"] = X_bits.sum(axis=1)

# -----------------------------------
# 2. Separate original train/test
# -----------------------------------

train_df = df[df["split"] == "train"]
test_df = df[df["split"] == "test"]

X_train = train_df[["hamming_weight"]]
y_train = train_df["label"].astype(int)

X_test = test_df[["hamming_weight"]]
y_test = test_df["label"].astype(int)

# -----------------------------------
# 3. Train using ONLY Hamming Weight
# -----------------------------------

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

# -----------------------------------
# 4. Evaluate
# -----------------------------------

pred = model.predict(X_test)

print("Hamming Weight Only")
print("-------------------")
print("Accuracy:", accuracy_score(y_test, pred))
print("F1 Score:", f1_score(y_test, pred))
```
